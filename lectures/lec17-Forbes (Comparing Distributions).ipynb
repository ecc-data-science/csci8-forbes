## Total Variation Distance

The total variation distance between two distributions measures how close the distributions are. 
- For each category, compute the difference in proportions between two distributions.
- Take the absolute value of each difference
- Sum, and then divide the sum by 2. 

```python
# Total Variation Distance

def tvd(dist1, dist2):    # dsit 1 and dist 2 are arrays of proportions 
    return sum(abs(dist1 - dist2))/2
```
