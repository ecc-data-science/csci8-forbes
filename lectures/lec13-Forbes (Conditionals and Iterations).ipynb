```python
# If statements are most commonly found in functions
# This is not a good version. 

def age(a):
    if a >= 18:
        return 'You can legally vote.'
    elif a >= 21:
        return 'You can legally vote and drink.'
```

```python
age(3)
```

```python
age(20)
```

```python
# The function doesn't get to 'You can legally drink'
age(40)
```

```python
# A better version

def age(a):
    if a >= 21:
        return 'You can legally vote and drink.'
    elif a >= 18: # This only gets checked if the if condition was False. 
        return 'You can legally vote.'
    else: # This only runs if all the if/elif conditions were False. 
        return 'You can legally drink milk.'
```

```python
age(3)
```

```python
age(20)
```

```python
age(23)
```
