# Models

- A model is a set of assumptions about the data
    - In data science, many models involve assumptions about processes that involve randomness
        - “Chance models”
- Key question: does the model fit the data?

**Steps in assessing a model**

- Make **assumptions** (model)
- **Choose a test statistic** to measure discrepancy between model and data
- **Simulate the statistic** under the model’s assumptions
- **Compare** the observed data to the model’s predictions:
    - Draw a histogram of simulated values of the statistic
    - Compute the observed statistic from the real sample
- If the observed statistic is far from the histogram, that is evidence against the model

**Case studies**

- Swain vs. Alabama
- Mendel's pea flowers

## Mendel and Pea Flowers ##

- Model: Each plant has a 75% chance of having purple flowers
- Test statistic: | sample percent of purple flower plants - 75 |

```python
# Mendel had 929 plants, of which 709 had purple flowers
# Compute the observed proportion of purple flowers. 

observed_purples = ...
observed_purples
```

```python
# Observed statistic (distance between observed proportion of purple flowers and 75%)

observed_statistic = ...
observed_statistic
```

```python
# Population   
predicted_proportions = make_array(...)         
# Sample
predicted_proportion_purple = ...        
# Test Statistic 
abs(predicted_proportion_purple - predicted_proportions.item(0)) * 100                
```

```python
# Generate sample proportions
# Return distance between percentage of purple-flowering plants from sample of 929 and 75%
# This is our statistic

def one_simulated_distance():
    predicted_proportion_purple = sample_proportions(929, predicted_proportions).item(0)
    return abs(predicted_proportion_purple - predicted_proportions.item(0)) * 100
```

```python
one_simulated_distance()
```

```python
# Simulate statistic 10000 times
# Add statistics to collection array

distances = make_array()

for i in np.arange(10000):
    new_simulated_distance = one_simulated_distance()
    distances = np.append(distances, new_simulated_distance)
distances
```

```python
# Distribution of the statistic -- distance from 75%

Table().with_column('Discrepancy in sample of 929 if the model is true', distances).hist()
plots.ylim(-0.02, 0.5)
plots.scatter(observed_statistic, 0, color = 'red', s = 40);

# Observed statistic seems to lie within the histogram
# Thus, the model seems consistent with the data
```
