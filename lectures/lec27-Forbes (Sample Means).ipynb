```python
import matplotlib
from datascience import *
%matplotlib inline
import matplotlib.pyplot as plots
import numpy as np
plots.style.use('fivethirtyeight')
```

## Example: United Airlines Flight Delays

```python
# Import United Airlines flight data, summer 2015
# Population distribution of delays

united = Table.read_table('united.csv')
united_bins = np.arange(-20, 300, 10)
united.hist('Delay', bins=united_bins)
```

```python
# Compute one sample mean

def one_sample_mean(sample_size):
    """Take a sample from the population of flights and compute its mean"""
    sampled_flights = united.sample(sample_size)
    return np.mean(sampled_flights.column('Delay'))
```

## Distribution of the Sample Mean

```python
# Compute 10,000 sample means

def ten_thousand_sample_means(sample_size):
    means = make_array()
    for i in np.arange(10000):
        mean = one_sample_mean(sample_size)
        means = np.append(means, mean)
    return means
```

```python
sample_means_400 = ten_thousand_sample_means(400)
Table().with_column('Mean of 400 flight delays', sample_means_400).hist(bins=20)
```

## Center of Distribution of the Sample Mean

```python
# Population Mean
delay_mean = np.average(united.column('Delay'))

# Mean of the distribution of 10000 sample means (n = 400)
sample_means_400 = ten_thousand_sample_means(400)
mean_sample_means = np.average(sample_means_400)

# Distribution of 10000 sample means (n = 400)
Table().with_column('Mean of 400 flight delays', sample_means_400).hist(bins=20)
print('Population Average:', delay_mean)
print('Average of Sample Averages',mean_sample_means)
```

## Variability of the Sample Mean

```python
sample_means_900 = ten_thousand_sample_means(900)
```

```python
means_tbl = Table().with_columns(
    '400', sample_means_400,
    '900', sample_means_900,
)
delay_sd = np.std(united.column('Delay'))
sample_means_400_sd = np.std(sample_means_400)
sample_means_900_sd = np.std(sample_means_900)
```

```python
means_tbl.hist(bins = np.arange(5, 31, 0.5))
plots.title('Distribution of Sample Average');
print("Population SD: ", delay_sd)
print("SD of sample averages, n = 400: ", sample_means_400_sd)
print("SD of sample averages, n = 900: ", sample_means_900_sd)
```
