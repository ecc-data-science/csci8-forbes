## Probability Distributions ##

- **Probability Distribution**:
    - A mathematical description that assigns probabilities to the possible outcomes of a random variable.
- **Theoretical Probability distribution**:
    - All possible unique values of the random variable
    - The theoretical (model-based) probability of each value
- **Empirical Probability Distribution**: 
    - Based on observed data (and may change from sample to sample)
    - All observed unique values
    - The observed relative frequency (proportion) of each value

```python
# Theoretical Probability distribution

die = Table().with_column('Face', np.arange(1, 7))
roll_bins = np.arange(0.5, 6.6, 1)
die.hist(bins=roll_bins)
```

```python
# Empirical Probability Distribution (sample size = 10)

...
```

```python
# Empirical probability distribution (sample size = 1000)
# Notice, distribution is still changing,
# but it looks more similar to the theoretical probability distribution

die.sample(1000).hist(bins=roll_bins)
```

```python
# Empirical probability distribution (sample size = 100000)
# This looks even more similar to the theoretical probability distribution

die.sample(100000).hist(bins=roll_bins)
```

## Parameters & Statistics ##

- **Parameter**: A numerical quantity associated with a population (this is a fixed number)
- **Statistic**: A numerical quantity calculated from a sample (this value varies depending on the random sample)

A statistic can be used as an estimate of a parameter to make some conclusion about a population.

### Probability & Empirical Distributions of a Statistic ###

**Sampling Distribution of a Statistic** is the probability distribution of a statistic (such as the mean, median, proportion) when calculated from all possible samples of a given sample size.
    - All possible values of the statistic,
    - and all the corresponding probabilities

**Empirical distribution of a statistic**
    - Based on simulated values of the statistic
    - Consists of all the observed values of the statistic,
    - and the proportion of times each value appeared
- Good approximation to the probability distribution of the statistic 
    - if the number of repetitions in the simulation is large

```python
# Population parameter: population median

np.median(united.column('Delay'))
```

```python
# This function samples data and returns the median

def sample_median(size):
    return np.median(united.sample(size).column('Delay'))
```

```python
# Simulating a sample median for sample size 10

sample_median(10)
```

#### Empirical Distributions of a Statistic (Overlayed)

```python
sample_medians_10 = make_array()
sample_medians_100 = make_array()
sample_medians_1000 = make_array()

num_simulations = 2000

for i in np.arange(num_simulations):
    new_median_10 = sample_median(10)
    sample_medians_10 = np.append(sample_medians_10, new_median_10)
    new_median_100 = sample_median(100)
    sample_medians_100 = np.append(sample_medians_100, new_median_100)
    new_median_1000 = sample_median(1000)
    sample_medians_1000 = np.append(sample_medians_1000, new_median_1000)
```

```python
sample_medians = Table().with_columns('Size 10', sample_medians_10, 
                                      'Size 100', sample_medians_100,
                                      'Size 1000', sample_medians_1000)
```

```python
sample_medians.hist(bins = np.arange(-5, 30))


# As sample size increases:
# 1) the sample statistic (e.g., sample median) becomes less variable; the sampling distribution narrows.
# 2) the distribution of the statistic clusters more closely around the population parameter, 
# resulting in a more accurate and reliable estimate.
```
